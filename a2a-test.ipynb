```python
import httpx
from a2a.client import A2ACardResolver
import uuid
from a2a.client import A2AClient
from a2a.types import (
    SendMessageRequest,
    SendMessageResponse,
    MessageSendParams
)
```

```python
client = httpx.AsyncClient(timeout=300)
card_resolver = A2ACardResolver(client, "http://localhost:8080", "/internet-search-agent/.well-known/a2a")
card = await card_resolver.get_agent_card()
a2a_client = A2AClient(client, card, url="http://localhost:8080/internet-search-agent")
print(card)
```

```python
task_id = str(uuid.uuid4())
context_id = str(uuid.uuid4())
message_id = str(uuid.uuid4())
```

```python
payload = {
    'message': {
        'role': 'user',
        'parts': [
            {'type': 'text', 'text': "What is graph rag"}
        ],  # Use the 'task' argument here
        'messageId': message_id,
    }
}
```

```python
req = SendMessageRequest(
    id = message_id,
    params = MessageSendParams.model_validate(payload)
)
```

```python
res = await a2a_client.send_message(req)
print(res)
```
